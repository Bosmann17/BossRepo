### Question B1: Linear Regression

The **diabetes** data set built into scikit-learn records, for 442 patients, ten measurements taken at a first visit (age, sex, body mass index, blood pressure, and size blood tests),  and a measure $y$ of how far the disease had progressed one year later. Load it and split it into a training set (80%) and text set (20%) where column 2 of **X** is the body ass index (BMI), in kg/$\textrm{m}^2$.

(a) **One Feature (Section 6)**

Using the training set only, fit $y=\beta_0+\beta_1\times\textrm{BMI}$ by writing down the design matrix $A$ and solving the normal equations $A^TA\beta=A^Ty$ yourselves (np.linalg.solve). Report $\hat{\beta}_0,\hat{\beta}_1, \textrm{and} r^2$ and $\hat{\beta}$ against scikit-learn's **LinearRegression**, and plot the training data with the fitted line. What does $\hat{\beta}_1$ mean in words?

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

X, y = load_diabetes(return_X_y=True, scaled=False)
X_tr, X_te, y_tr, y_te = train_test_split(X, y, test_size=0.2, random_state=0)
```

## (a) BMI-only regression (2 marks)

The model is $\hat y=\hat\beta_0+\hat\beta_1\,\mathrm{BMI}$. Using the training data, the design matrix and normal equations are

$$A=\begin{bmatrix}1 & x_1\\1 & x_2\\\vdots & \vdots\\1 & x_{353}\end{bmatrix},\qquad A^TA\hat\beta=A^Ty_{tr}.$$

We solve for $\hat\beta$ using `np.linalg.solve` and compare the result with `LinearRegression`.

```python
bmi_tr = X_tr[:, 2]
A = np.column_stack((np.ones(len(bmi_tr)), bmi_tr))
beta_0, beta_1 = np.linalg.solve(A.T @ A, A.T @ y_tr)
r2 = r2_score(y_tr, A @ np.array([beta_0, beta_1]))

bmi_model = LinearRegression().fit(X_tr[:, [2]], y_tr)
print(f"Normal equations: intercept = {beta_0:.4f}, BMI coefficient = {beta_1:.4f}")
print(f"Training R² = {r2:.4f}")
print(f"Scikit-learn:     intercept = {bmi_model.intercept_:.4f}, BMI coefficient = {bmi_model.coef_[0]:.4f}")

x_line = np.linspace(bmi_tr.min(), bmi_tr.max(), 200)
plt.figure(figsize=(8, 5))
plt.scatter(bmi_tr, y_tr, alpha=0.5, label="Training data")
plt.plot(x_line, beta_0 + beta_1 * x_line, color="red", label="Fitted line")
plt.xlabel("BMI (kg/m²)")
plt.ylabel("Diabetes progression")
plt.title("BMI and Diabetes Progression")
plt.legend()
plt.tight_layout()
plt.show()
```

The fitted model is $\hat y=-126.7782+10.5804(\mathrm{BMI})$, with a training $R^2$ of 0.3756. This means BMI explains about 37.56% of the variation in diabetes progression in the training set. The coefficients match scikit-learn, and the BMI slope means that a one-unit increase in BMI is associated with about 10.58 more units of predicted progression.

## (b) All ten features and model comparison (2 marks)

We compare the training and test mean squared errors, where $\mathrm{MSE}=\frac{1}{n}\sum_{i=1}^n(y_i-\hat y_i)^2$. The baseline predicts the training mean for every test patient.

```python
full_model = LinearRegression().fit(X_tr, y_tr)

bmi_train_mse = mean_squared_error(y_tr, bmi_model.predict(X_tr[:, [2]]))
bmi_test_mse = mean_squared_error(y_te, bmi_model.predict(X_te[:, [2]]))
full_train_mse = mean_squared_error(y_tr, full_model.predict(X_tr))
full_test_mse = mean_squared_error(y_te, full_model.predict(X_te))
baseline_test_mse = mean_squared_error(y_te, np.full(len(y_te), y_tr.mean()))

print(f"{'Model':<20} {'Training MSE':>14} {'Test MSE':>12}")
print(f"{'BMI only':<20} {bmi_train_mse:>14.2f} {bmi_test_mse:>12.2f}")
print(f"{'All ten features':<20} {full_train_mse:>14.2f} {full_test_mse:>12.2f}")
print(f"{'Training-mean baseline':<20} {'—':>14} {baseline_test_mse:>12.2f}")
```

The ten-feature model performs best on the test set, with an MSE of 3424.26, compared with 4150.68 for BMI alone and 5134.78 for the baseline. Its training MSE must be no higher than the BMI-only model’s because it could set the other nine coefficients to zero and produce the same predictions. However, a lower training MSE does not necessarily mean better predictions for new patients, since extra features can cause overfitting. This is why we compare test MSE.
